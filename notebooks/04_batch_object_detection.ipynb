```python
import sys
sys.path.insert(0, "../src")

import pandas as pd
from batch_object_pipeline import run_batch_object_detection

FRAMES_ROOT = "../outputs/sampled_frames"
OUT_ROOT = "../outputs/object_detections"

MODEL_NAME = "yolov8n.pt"   # fast & small; good for Mac
CONF = 0.25
IOU = 0.5

SAVE_ANNOTATED = True
MAX_ANN_PER_VIDEO = 8

ENABLE_SMALL_OBJECT_PROPOSALS = True
```

```python
summary_df = run_batch_object_detection(
    frames_root_dir=FRAMES_ROOT,
    out_root_dir=OUT_ROOT,
    model_name=MODEL_NAME,
    conf=CONF,
    iou=IOU,
    save_annotated_examples=SAVE_ANNOTATED,
    max_annotated_examples_per_video=MAX_ANN_PER_VIDEO,
    enable_small_object_proposals=ENABLE_SMALL_OBJECT_PROPOSALS
)

summary_df.to_csv("../outputs/object_detection_summary.csv", index=False)
summary_df.head()
```
