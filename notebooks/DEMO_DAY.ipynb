```python
import sys
sys.path.insert(0, "../src")

import json
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from pathlib import Path
from graph_visualizer import (
    visualize_interactive,
    query_graph_by_skill,
    get_cross_domain_skills
)
from manual_to_skill_graph import (
    build_manual_skill_graph,
    merge_manual_graph_into_master
)

# Load master graph
with open("../outputs/skill_graphs/DRISHTI_master_graph.json") as f:
    data = json.load(f)

G = nx.DiGraph()
for node in data["nodes"]:
    nid = node.pop("id")
    G.add_node(nid, **node)
for edge in data["edges"]:
    src = edge.pop("source")
    tgt = edge.pop("target")
    G.add_edge(src, tgt, **edge)

print("✅ DRISHTI Ready for Demo")
print(f"   Nodes: {G.number_of_nodes()} | Edges: {G.number_of_edges()}")
```

```python
enriched_files = list(Path("../outputs/robot_skills").rglob("*_enriched.csv"))
dfs = [pd.read_csv(f) for f in enriched_files if f.stat().st_size > 0]
combined = pd.concat(dfs, ignore_index=True)

sample = combined.iloc[0].to_dict()

print("=" * 62)
print("🎓 SAMPLE OUTPUT — One Complete Intelligence Node")
print("=" * 62)
print(f"📹 Video: {sample.get('video_stem')}")
print(f"🏷️  Domain: {sample.get('domain')}")
print(f"⏱️  Segment Frames: {sample.get('start_frame')} → {sample.get('end_frame')}")
print()
print("┌─────────────────────────────────────────────────────┐")
print(f"│  🤲 KINEMATIC  : {str(sample.get('grasp_used','')):<36}│")
print(f"│  🎯 INTENT     : {str(sample.get('intent','')):<36}│")
print(f"│  🤖 ROBOT SKILL: {str(sample.get('robot_skill','')):<36}│")
print("├─────────────────────────────────────────────────────┤")
print(f"│  📝 {str(sample.get('intent_description',''))[:52]:<53}│")
print(f"│  ⚙️  Complexity : {str(sample.get('skill_complexity','')):<36}│")
print(f"│  🎮 DOF        : {str(sample.get('dof_required','')):<36}│")
print(f"│  👐 Bimanual   : {str(sample.get('bimanual','')):<36}│")
print("└─────────────────────────────────────────────────────┘")
```

```python
cross = get_cross_domain_skills(G)

print("=" * 62)
print("🔗 CROSS-DOMAIN SKILL TRANSFER")
print("=" * 62)
print()
print("Skills that appear across MULTIPLE domains:")
print("(This is what enables transfer learning in robots)\n")

for skill, domains in cross.items():
    print(f"  ✨ {skill}")
    print(f"     → Domains: {' + '.join(domains)}")
    print()

print(f"\n💡 Total transferable skills: {len(cross)}")
print("""
What this means for robotics:
→ A robot trained on jewellery tasks can REUSE precision 
  skills for shop or cleaning tasks.
→ No need to retrain from scratch for every new domain.
→ This is the foundation of Hierarchical RL (CoRL-style).
""")
```

```python
print("🔍 SKILL QUERY ENGINE DEMO")
print("=" * 62)
print("Query: Find all segments requiring 'dexterous_fine_manipulation'")
print()

results = query_graph_by_skill(G, "dexterous_fine_manipulation")
df_query = pd.DataFrame(results)

if not df_query.empty:
    display(df_query[["domain", "video", "frames", "intent", "confidence"]])
    print(f"\nTotal matches: {len(df_query)}")
    print("""
What a robot can do with this:
→ Build a curriculum: learn fine manipulation first
→ Identify which videos to imitate for this specific skill
→ Estimate how many examples exist per skill
""")
```

```python
print("=" * 62)
print("⭐ BONUS FEATURE: TEXT MANUAL → SKILL GRAPH")
print("=" * 62)
print()

# Example: Jewellery Making SOP (could be any domain manual)
jewellery_manual = """
Step 1: Pick up the beading wire from the tray.
Step 2: Thread the first bead onto the wire carefully.
Step 3: Inspect the bead for any defects or cracks.
Step 4: Place the threaded bead at the end of the wire.
Step 5: Pick up the next bead and thread it onto the wire.
Step 6: Arrange the beads evenly along the wire.
Step 7: Fold the wire ends to secure the beads.
Step 8: Inspect the final necklace for quality.
Step 9: Place the finished necklace in the packaging box.
"""

print("📄 INPUT MANUAL (Jewellery Making SOP):")
print(jewellery_manual)

manual_G, steps = build_manual_skill_graph(
    jewellery_manual, 
    domain="manual_jewellery",
    title="Jewellery Making SOP"
)

print(f"\n✅ Manual graph built!")
print(f"   Nodes: {manual_G.number_of_nodes()}")
print(f"   Steps parsed: {len(steps)}")

# Show extracted skills
print("\n📊 Skills extracted from manual:")
for node, attrs in manual_G.nodes(data=True):
    if attrs.get("node_type") == "manual_step":
        print(f"  Step {attrs['step_number']}: {attrs['robot_skill']} "
              f"(from verb: '{attrs['action_verb']}')")
```

```python
print("=" * 62)
print("🔗 MERGING MANUAL GRAPH INTO MASTER SKILL GRAPH")
print("=" * 62)

extended_G = merge_manual_graph_into_master(G, manual_G)

# Save extended graph
extended_path = "../outputs/DRISHTI_EXTENDED_master_graph.json"
from skill_graph_builder import export_graph_to_json_ld
export_graph_to_json_ld(extended_G, extended_path)

# Generate interactive visualization of extended graph
from graph_visualizer import visualize_interactive
visualize_interactive(
    extended_G,
    "../outputs/visualizations/extended_graph_with_manual.html",
    color_by="skill"
)

print(f"""
✅ Extended graph created!
   Original nodes: {G.number_of_nodes()}
   After adding manual: {extended_G.number_of_nodes()}
   
💡 What this means:
→ DRISHTI can learn from BOTH video AND text manuals
→ A construction SOP → robot skill graph → robot learns the job
→ This is exactly what the judge mentioned from CoRL work!
→ No video needed — just give us any procedural text

Open: outputs/visualizations/extended_graph_with_manual.html
""")
```
